# Mission 4：算術運算與成績公式

Knowledge Point

* 基本算術運算子：+（加）、-（減）、*（乘）、/（除）。
* 可以先分步驟計算各項影響，最後再加總，讓程式更易閱讀。
* 此處的公式僅為示例，並不代表真實因果關係。

Markdown Cell

我們以一位假想學生的讀書時數、睡眠時數、螢幕使用時數與出席率，逐步計算模擬的小考成績。

```python
# 假想學生的基本資料
study_hours = 2.5          # 讀書時數（小時）
sleep_hours = 7.0          # 睡眠時數（小時）
screen_time_hours = 4.0    # 螢幕使用時數（小時）
attendance_rate = 90       # 出席率（%）

# 分別計算各項對成績的影響
base_score = 50
study_effect = 3 * study_hours
sleep_effect = 2 * sleep_hours
screen_effect = -1.2 * screen_time_hours   # 負數代表扣分
attendance_effect = 0.15 * attendance_rate

# 加總得到未加隨機誤差的成績
quiz_score_raw = base_score + study_effect + sleep_effect + screen_effect + attendance_effect

print("基礎分數:", base_score)
print("讀書時數影響:", study_effect)
print("睡眠時數影響:", sleep_effect)
print("螢幕使用時數影響:", screen_effect)
print("出席率影響:", attendance_effect)
print("未加隨機誤差的成績:", quiz_score_raw)
```

Markdown Cell

建立一個簡單的函式 describe_score，依照分數回傳對應的評語。

```python
def describe_score(score):
    """根據分數回傳評語"""
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

# 測試不同分數
test_scores = [92, 75, 58, 40]
for s in test_scores:
    print(f"分數 {s} => {describe_score(s)}")
```

# Mission 7：建立模擬的大學生學習資料

Knowledge Point

* 使用 pandas.DataFrame 建立表格資料，每列代表一位學生，每欄代表一個變數。
* np.random.default_rng().normal(mean, std, size) 產生常態分布隨機數。
* .clip(a_min, a_max) 將數值限制在指定範圍內。
* round(1) 表示保留一位小數；astype(int) 轉成整數。

Markdown Cell

現在我們將根據之前設定的參數，產生 60 筆模擬大學生學習資料，並放入變數 df 中。

```python
# 學生總數（先前已定義）
n = student_count   # 60

# 使用相同的亂數種子，確保可重現
rng = np.random.default_rng(random_seed)

# 建立 DataFrame
df = pd.DataFrame({
    "student_id": np.arange(1, n+1),                                   # 1~60
    "department": rng.choice(departments, size=n),                     # 隨機選擇科系
    "sleep_hours": rng.normal(loc=average_sleep_hours, scale=1.1, size=n).round(1),
    "screen_time_hours": rng.normal(loc=5.0, scale=1.5, size=n).clip(1, 10).round(1),
    "study_hours":   rng.normal(loc=2.5, scale=1.0, size=n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(loc=88, scale=8, size=n).clip(60, 100).astype(int)
})

# 顯示前五筆資料
df.head()
```

# Mission 8：建立小考成績欄位

Knowledge Point

* 可以直接對 DataFrame 欄位進行向量化運算（例如 df["study_hours"] * 3）。
* np.random.default_rng().normal(0, 5, size=n) 產生平均 0、標準差 5 的隨機誤差。
* .clip(0, 100) 將成績限制在 0~100 分之間。
* .round().astype(int) 四捨五入成整數。

Markdown Cell

根據之前的公式，為每位學生計算 quiz_score（小考成績），並加入隨機誤差。

```python
# 取得各欄位的 NumPy 陣列（方便向量運算）
study   = df["study_hours"].to_numpy()
sleep   = df["sleep_hours"].to_numpy()
screen  = df["screen_time_hours"].to_numpy()
attend  = df["attendance_rate"].to_numpy()

# 各項影響（依照公式分開計算）
base = 50
study_effect   = 3 * study
sleep_effect   = 2 * sleep
screen_effect  = -1.2 * screen
attend_effect  = 0.15 * attend

# 隨機誤差（平均 0，標準差 5）
rng = np.random.default_rng(random_seed)   # 保持種子一致
error = rng.normal(loc=0.0, scale=5.0, size=n)

# 原始成績（未限制）
quiz_raw = base + study_effect + sleep_effect + screen_effect + attend_effect + error

# 限制在 0~100，四捨五入成整數
df["quiz_score"] = quiz_raw.clip(0, 100).round().astype(int)

# 檢查前五筆資料（含我們關心的欄位）
df[["student_id","study_hours","sleep_hours","screen_time_hours","attendance_rate","quiz_score"]].head()
```

# Mission 11：篩選與比較資料

Knowledge Point

* 單欄選取：df["column_name"] 回傳 Series。
* 多欄選取：df[["col1","col2"]] 回傳新 DataFrame。
* 布林條件：例如 df["quiz_score"] >= 80 會產生一個 True/False 系列，可用來篩選列。
* 為了易於閱讀，先把條件儲存變數再用來索引。

Markdown Cell

我們示範如何：選取單一欄位、多個欄位、找出缺失值、找出螢幕使用時間超過 10 小時的學生，以及找出成績 80 分以上的學生。

```python
# 1️⃣ 單一欄位（例如 quiz_score）
print("=== 單一欄位 quiz_score 前五筆 ===")
print(df["quiz_score"].head())

# 2️⃣ 多個欄位
print("\n=== 多個欄位前五筆 ===")
print(df[["student_id","quiz_score"]].head())

# 3️⃣ 找出有缺失值的資料（以 sleep_hours 為例）
missing_sleep = df["sleep_hours"].isna()
print("\n=== 有缺失值的 sleep_hours 列索引 ===")
print(df[missing_sleep][["student_id","sleep_hours"]])

# 4️⃣ 螢幕使用時間 > 10 小時
high_screen = df["screen_time_hours"] > 10
print("\n=== 螢幕使用時間 > 10 小時的學生 ===")
print(df[high_screen][["student_id","screen_time_hours"]])

# 5️⃣ 成績 80 分以上
high_score = df["quiz_score"] >= 80
print("\n=== 成績 80 分以上的學生 ===")
print(df[high_score][["student_id","quiz_score"]].head())
```

# 最終整合任務

以下是一個 可從頭到尾直接執行 的完整程式，涵蓋所有前面 Missions 的步驟：匯入套件、設定亂數種子、建立模擬資料、加入 quiz_score、故意加入缺失值與異常值、顯示前五筆、做基本資料檢查、匯出 CSV。

* 執行前請確認：如果你在 Google Colab 中執行，最後兩行的下載程式才會起作用；在本機 Jupyter 中，檔案會直接寫到目前工作目錄，無需下載指令。

```python
# ==============================
# 最終整合任務：從匯入到匯出 CSV
# ==============================

# 1️⃣ 套件匯入
import numpy as np
import pandas as pd

# 2️⃣ 基本參數（與前面任務一致）
student_count = 60
average_sleep_hours = 6.7
course_name = "Python與資料科學"
has_missing_value = True   # 此變數未在後續直接使用，僅作為示例

# 3️⃣ 可重現的亂數種子
random_seed = 42
rng = np.random.default_rng(random_seed)

# 4️⃣ 科系串列（用於隨機抽樣）
departments = ["資訊管理", "企業管理", "金融管理"]

# 5️⃣ 建立模擬資料 DataFrame
n = student_count
df = pd.DataFrame({
    "student_id": np.arange(1, n+1),
    "department": rng.choice(departments, size=n),
    "sleep_hours":   rng.normal(loc=average_sleep_hours, scale=1.1, size=n).round(1),
    "screen_time_hours": rng.normal(loc=5.0, scale=1.5, size=n).clip(1, 10).round(1),
    "study_hours":   rng.normal(loc=2.5, scale=1.0, size=n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(loc=88, scale=8, size=n).clip(60, 100).astype(int)
})

# 6️⃣ 計算 quiz_score（依照公式，含隨機誤差）
study   = df["study_hours"].to_numpy()
sleep   = df["sleep_hours"].to_numpy()
screen  = df["screen_time_hours"].to_numpy()
attend  = df["attendance_rate"].to_numpy()

base = 50
study_effect   = 3 * study
sleep_effect   = 2 * sleep
screen_effect  = -1.2 * screen
attend_effect  = 0.15 * attend

error = rng.normal(loc=0.0, scale=5.0, size=n)   # 平均 0，標準差 5
quiz_raw = base + study_effect + sleep_effect + screen_effect + attend_effect + error

df["quiz_score"] = quiz_raw.clip(0, 100).round().astype(int)

# 7️⃣ 故意加入缺失值與異常值（與前面任務相同）
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0

# 8️⃣ 檢查前五筆資料（含我們關心的欄位）
print("=== 前五筆資料 ===")
df[["student_id","study_hours","sleep_hours","screen_time_hours","attendance_rate","quiz_score"]].head()

# 9️⃣ 基本資料品質檢查
print("\n=== 資料形狀 (shape) ===")
print(df.shape)

print("\n=== 欄位型態 (dtypes) ===")
print(df.dtypes)

print("\n=== 缺失值數量 (isna().sum()) ===")
print(df.isna().sum())

print("\n=== 敘述統計 (describe) ===")
print(df.describe())

# 🔟 匯出 CSV 檔案
file_name = "student_learning_data.csv"
df.to_csv(
    file_name,
    index=False,
    encoding="utf-8-sig"
)
print(f"\n資料已儲存為：{file_name}")

# 只有在 Google Colab 才會觸發下載
try:
    from google.colab import files
    files.download(file_name)
    print("（Colab 環境）檔案已觸發下載。")
except Exception:
    # 非 Colab 環境，不做任何事
    pass
```

執行完成後，你應該會看到：

* 前五筆資料的表格印出。
* 資料形狀為 (60, 8)（包括 student_id 到 quiz_score 共 8 欄）。
* sleep_hours 欄位顯示 2 筆缺失值。
* screen_time_hours 欄位的最大值遠大於 10（因為我們故意設了 15）。
* 工作目錄中出現 student_learning_data.csv 檔案。
